```python
# Import required libraries
import sqlite3
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Connect to the SQLite database
conn = sqlite3.connect("sales_data.db")
cursor = conn.cursor()
```

```python
# Create a 'sales' table if it doesn't exist
cursor.execute("""
CREATE TABLE IF NOT EXISTS sales (
    id INTEGER PRIMARY KEY AUTOINCREMENT,
    product TEXT,
    quantity INTEGER,
    price REAL
)""")
```

```python
# Insert some dummy sales data
sample_data = [
    ('Apples', 10, 2.5),
    ('Bananas', 5, 1.2),
    ('Oranges', 8, 1.8),
    ('Apples', 7, 2.5),
    ('Bananas', 12, 1.2),
    ('Oranges', 4, 1.8),
]

cursor.executemany("INSERT INTO sales (product, quantity, price) VALUES (?, ?, ?)", sample_data)
conn.commit()
```

```python
# SQL query to get total quantity and revenue by product
query = """
SELECT
    product,
    SUM(quantity) AS total_qty,
    SUM(quantity * price) AS revenue
FROM sales
GROUP BY product
"""
```

```python
# Use pandas to run the query and load the result into a DataFrame
df = pd.read_sql_query(query, conn)

# Step 7: Print the result
print("Sales Summary:")
print(df)
```

```python
# Plot a bar chart showing revenue by product
df.plot(kind='bar', x='product', y='revenue', color='skyblue', legend=False)
plt.title("Revenue by Product")
plt.xlabel("Product")
plt.ylabel("Revenue")
plt.tight_layout()
```

```python
# Save the chart as an image
plt.savefig("sales_chart.png")
plt.show()
```

```python
# Close the database connection
conn.close()
```
